```python
import pandas as pd
import matplotlib.pyplot as plt

kinetics=pd.DataFrame({'Time_min':[0,10,20,30,40,50],'Run1_mol_L':[1.2,0.98,0.80,0.66,0.54,0.44],'Run2_mol_L':[1.20,0.8,0.54,0.36,0.24,0.16]})
kinetics
```

```python
x=kinetics.to_csv('kinetics_data.csv', index=False)
y=kinetics.to_json('kinetics_data.json', orient='records')

kinetics_csv=pd.read_csv('kinetics_data.csv')
print(kinetics_csv.shape)
kinetics_csv.dtypes
```

```python
std1=kinetics_csv['Run1_mol_L'].std()
mean1=kinetics_csv['Run1_mol_L'].mean()
print(f'For Run1 the mean is {mean1} and the standard deviation is {std1}')

std2=kinetics_csv['Run2_mol_L'].std()
mean2=kinetics_csv['Run2_mol_L'].mean()
print(f'For Run2 the mean is {mean2} and the standard deviation is {std2}')
```

```python
plt.figure(figsize=(8,6))
plt.plot(kinetics_csv['Time_min'], kinetics_csv['Run1_mol_L'], marker='o', label='Run 1', color='b')
plt.plot(kinetics_csv['Time_min'], kinetics_csv['Run2_mol_L'], marker='s', label='Run 2', color='r')
plt.title('Decomposition of Hydrogen peroxide')
plt.xlabel('Time_min')
plt.ylabel('Concentration (mol/L)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
carbon=pd.DataFrame({'volume_L':[1,2,3,4,5],'Pressure_bar':[3.02,1.49,1.01,0.76,0.59]})
carbon
```

```python
carbon['PxV']=carbon['Pressure_bar']*carbon['volume_L']
carbon
k=carbon['PxV'].mean()
print(f'The mean of PxV is: {k} bar.L')
```

```python
plt.figure(figsize=(8,6))
plt.scatter(carbon['volume_L'], carbon['Pressure_bar'], marker='o', label='Pressure vs Volume', color='b')
plt.scatter(carbon['volume_L'], k/carbon['volume_L'], marker='s', label='Pressure(P=k/V) vs Volume', color='r')
plt.xlabel('Volume (L)')
plt.ylabel('Pressure (bar)')
plt.title('Pressure vs Volume for Carbon dioxide')
plt.legend()
plt.grid()
plt.show()

```
